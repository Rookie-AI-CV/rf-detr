```python
from torchinfo import summary
```

```python
from rfdetr import RFDETRLarge

model = RFDETRLarge(
    pretrain_weights="/root/autodl-tmp/model/rf-detr-large.pth",
)
```

```python
import torch
from torchviz import make_dot

# Create input tensor
x = torch.randn(1, 3, 1008, 1008)
# 提取字典中的单个张量作为可视化对象
output_dict = model.model.model(x)
tensor_to_visualize = list(output_dict.values())[0]  # 获取第一个输出张量

# 创建可视化
dot = make_dot(tensor_to_visualize, params=dict(model.model.model.named_parameters()))
dot.format = 'png'
dot.render("model_architecture", cleanup=True)
```

```python
summary(model.model.model, input_size=(1, 3, 1008, 1008), device="cpu")
```
